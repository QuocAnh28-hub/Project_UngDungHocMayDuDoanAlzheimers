```python
import pandas as pd

path = '/content/drive/MyDrive/Học máy cơ bản/diabetes.csv'

df = pd.read_csv(path)

#1: Đọc file diabetes.csv và hiển thị 10 bản ghi đầu tiên.
df.head(10)
```

```python
#2: Hiển thị 10 bản ghi ngẫu nhiên trong file diabetes.csv.
df.sample(10)
```

```python
#3: Hiển thị số người bị mắc bệnh tiểu đường và số người không bị.
x = (df['Outcome'] == 1).sum()
y = (df['Outcome'] == 0).sum()
print("Số người bị mắc bệnh tiểu đường: ", x)
print("Số người không bị: ", y)
```

```python
#4: Hiển thị tỉ lệ phần trăm số người bị mắc bệnh và không mắc bệnh.
x = (df['Outcome'] == 1).sum()
y = (df['Outcome'] == 0).sum()
tong = x + y
x = (x / tong) * 100
y = (y / tong) * 100
print("Tỷ lệ % người bị mắc bệnh tiểu đường: ", x, "%")
print("Tỷ lệ % người không bị: ", y, "%")
```

```python
#5: Đếm và hiển thị số người trên 50 tuổi.
x = (df['Age']>50).sum()
print("Số người trên 50 tuổi: ", x)
df[df.Age>50]
```

```python
#6: Đếm và hiển thị số người trên 50 tuổi mà không bị mắc bệnh tiểu đường.
x = df[(df['Age']>50) & (df['Outcome'] == 0)]
print("Số người trên 50 tuổi không bị mắc tiểu đường: ", len(x))
x
```

```python
#7: Đếm và hiển thị số người trên 50 tuổi, mang thai ít nhất 2 lần mà không bị mắc bệnh tiểu đường.
x = df[(df['Age']>50) & (df['Outcome'] == 0) & (df['Pregnancies']>=2)]
print("Số người trên 50 tuổi mang thai ít nhất 2 lần không bị mắc tiểu đường: ", len(x))
x
```

```python
#8: Đếm xem có bao nhiêu người mang thai ít nhất 3 lần.
x = df[df['Pregnancies']>=3]
print("Số người mang thai ít nhất 3 lần: ", len(x))
x
```

```python
#9: Đếm xem có bao nhiêu người trên 40 tuổi và mang thai ít nhất 2 lần.
x = df[(df['Age']>40) & (df['Pregnancies']>=2)]
print("Số người trên 40 tuổi mang thai ít nhất 2 lần: ", len(x))
x
```
