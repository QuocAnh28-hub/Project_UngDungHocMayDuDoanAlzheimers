```python
# Yêu cầu 1: Đọc dữ liệu từ file iris.csv. Chia dữ liệu random thành 2 phần: training
# (85%) và phần còn lại cho testing (15%).
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# 1. Đọc dữ liệu từ file Iris.csv
df = pd.read_csv('/content/drive/MyDrive/Học máy cơ bản/Iris.csv')

# 2. Loại bỏ cột 'Id' và tách đặc trưng (X), nhãn (y)
X = df.drop(columns=['Id', 'Species']).values
y_raw = df['Species'].values

# 3. Mã hóa nhãn chữ thành số (0, 1, 2) phục vụ tính toán chỉ số
encoder = LabelEncoder()
y = encoder.fit_transform(y_raw)

# 4. Chia dữ liệu thành training (85%) và testing (15%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.15, random_state=42)

print("Số lượng mẫu huấn luyện (Train):", len(X_train))
print("Số lượng mẫu kiểm thử (Test):", len(X_test))
```

```python
# Yêu cầu 2: Huấn luyện, đánh giá và hiển thị giá trị: Loss, Accuracy, Precision,
# Recall với giải thuật K-Nearest Neighbor.
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, log_loss

# 1. Huấn luyện mô hình KNN
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)

# 2. Dự đoán trên tập testing
y_pred_knn = knn_model.predict(X_test)
y_pred_prob_knn = knn_model.predict_proba(X_test)

# 3. Tính toán các chỉ số đánh giá
knn_loss = log_loss(y_test, y_pred_prob_knn)
knn_acc = accuracy_score(y_test, y_pred_knn)
knn_prec = precision_score(y_test, y_pred_knn, average='macro')
knn_rec = recall_score(y_test, y_pred_knn, average='macro')

# 4. Hiển thị giá trị
print("--- K-NEAREST NEIGHBOR (KNN) ---")
print(f"Loss (Log-loss): {knn_loss:.4f}")
print(f"Accuracy:        {knn_acc:.4f}")
print(f"Precision:       {knn_prec:.4f}")
print(f"Recall:          {knn_rec:.4f}")
```

```python
# Yêu cầu 3: Huấn luyện, đánh giá và hiển thị giá trị: Loss, Accuracy, Precision,
# Recall với giải thuật Naïve Bayes.
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, log_loss

# 1. Huấn luyện mô hình Naïve Bayes
nb_model = GaussianNB()
nb_model.fit(X_train, y_train)

# 2. Dự đoán trên tập testing
y_pred_nb = nb_model.predict(X_test)
y_pred_prob_nb = nb_model.predict_proba(X_test)

# 3. Tính toán các chỉ số đánh giá
nb_loss = log_loss(y_test, y_pred_prob_nb)
nb_acc = accuracy_score(y_test, y_pred_nb)
nb_prec = precision_score(y_test, y_pred_nb, average='macro')
nb_rec = recall_score(y_test, y_pred_nb, average='macro')

# 4. Hiển thị giá trị
print("--- NAÏVE BAYES ---")
print(f"Loss (Log-loss): {nb_loss:.4f}")
print(f"Accuracy:        {nb_acc:.4f}")
print(f"Precision:       {nb_prec:.4f}")
print(f"Recall:          {nb_rec:.4f}")
```

```python
# Yêu cầu 4: Huấn luyện, đánh giá và hiển thị giá trị: Loss, Accuracy, Precision,
# Recall với giải thuật Support Vector Machine.
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, log_loss

# 1. Huấn luyện mô hình SVM (bật probability=True để tính được Log-loss)
svm_model = SVC(kernel='linear', probability=True, random_state=42)
svm_model.fit(X_train, y_train)

# 2. Dự đoán trên tập testing
y_pred_svm = svm_model.predict(X_test)
y_pred_prob_svm = svm_model.predict_proba(X_test)

# 3. Tính toán các chỉ số đánh giá
svm_loss = log_loss(y_test, y_pred_prob_svm)
svm_acc = accuracy_score(y_test, y_pred_svm)
svm_prec = precision_score(y_test, y_pred_svm, average='macro')
svm_rec = recall_score(y_test, y_pred_svm, average='macro')

# 4. Hiển thị giá trị
print("--- SUPPORT VECTOR MACHINE (SVM) ---")
print(f"Loss (Log-loss): {svm_loss:.4f}")
print(f"Accuracy:        {svm_acc:.4f}")
print(f"Precision:       {svm_prec:.4f}")
print(f"Recall:          {svm_rec:.4f}")
```

```python
# Yêu cầu 5: Huấn luyện, đánh giá và hiển thị giá trị: Loss, Accuracy, Precision,
# Recall với giải thuật Logistic Regression.
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, log_loss

# 1. Huấn luyện mô hình Logistic Regression
lr_model = LogisticRegression(max_iter=200, random_state=42)
lr_model.fit(X_train, y_train)

# 2. Dự đoán trên tập testing
y_pred_lr = lr_model.predict(X_test)
y_pred_prob_lr = lr_model.predict_proba(X_test)

# 3. Tính toán các chỉ số đánh giá
lr_loss = log_loss(y_test, y_pred_prob_lr)
lr_acc = accuracy_score(y_test, y_pred_lr)
lr_prec = precision_score(y_test, y_pred_lr, average='macro')
lr_rec = recall_score(y_test, y_pred_lr, average='macro')

# 4. Hiển thị giá trị
print("--- LOGISTIC REGRESSION ---")
print(f"Loss (Log-loss): {lr_loss:.4f}")
print(f"Accuracy:        {lr_acc:.4f}")
print(f"Precision:       {lr_prec:.4f}")
print(f"Recall:          {lr_rec:.4f}")
```

```python
# Yêu cầu 6: Huấn luyện, đánh giá và hiển thị giá trị: Loss, Accuracy, Precision,
# Recall với giải thuật Decision Tree.
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, log_loss

# 1. Huấn luyện mô hình Decision Tree
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)

# 2. Dự đoán trên tập testing
y_pred_dt = dt_model.predict(X_test)
y_pred_prob_dt = dt_model.predict_proba(X_test)

# 3. Tính toán các chỉ số đánh giá
dt_loss = log_loss(y_test, y_pred_prob_dt)
dt_acc = accuracy_score(y_test, y_pred_dt)
dt_prec = precision_score(y_test, y_pred_dt, average='macro')
dt_rec = recall_score(y_test, y_pred_dt, average='macro')

# 4. Hiển thị giá trị
print("--- DECISION TREE ---")
print(f"Loss (Log-loss): {dt_loss:.4f}")
print(f"Accuracy:        {dt_acc:.4f}")
print(f"Precision:       {dt_prec:.4f}")
print(f"Recall:          {dt_rec:.4f}")
```

```python
# Yêu cầu 7: So sánh với kết quả ở Bài 1, xem giải thuật nào cho kết quả tốt nhất.
import pandas as pd

# Giả định kết quả Bài 1 (Mạng nơ-ron Conv1D) thu được từ kết quả chạy thực tế của bạn
# Bạn hãy thay thế các giá trị bên dưới bằng chỉ số thực tế trên màn hình của bạn
nn_loss, nn_acc, nn_prec, nn_rec = 0.1500, 0.9565, 0.9600, 0.9500

# Tạo bảng so sánh
data = {
    "Giải thuật": ["Mạng Nơ-ron (Bài 1)", "KNN", "Naïve Bayes", "SVM", "Logistic Regression", "Decision Tree"],
    "Loss": [nn_loss, knn_loss, nb_loss, svm_loss, lr_loss, dt_loss],
    "Accuracy": [nn_acc, knn_acc, nb_acc, svm_acc, lr_acc, dt_acc],
    "Precision": [nn_prec, knn_prec, nb_prec, svm_prec, lr_prec, dt_prec],
    "Recall": [nn_rec, knn_rec, nb_rec, svm_rec, lr_rec, dt_rec]
}

df_compare = pd.DataFrame(data)
print("=== BẢNG SO SÁNH HIỆU NĂNG CÁC GIẢI THUẬT ===")
display(df_compare.round(4))

# Tìm giải thuật có Accuracy cao nhất
best_algo = df_compare.loc[df_compare['Accuracy'].idxmax()]['Giải thuật']
print(f"\n=> Kết luận: Giải thuật cho kết quả tốt nhất dựa trên độ chính xác (Accuracy) là: {best_algo}")
```
