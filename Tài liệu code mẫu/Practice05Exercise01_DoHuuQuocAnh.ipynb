```python
# Yêu cầu 1: Đọc dữ liệu từ file advertising.csv. Thực hiện:
#  -Hiển thị biểu đồ mối quan hệ giữa sales và TV (biểu đồ 1).
#  -Hiển thị biểu đồ mối quan hệ giữa sales và radio (biểu đồ 2).
#  -Hiển thị biểu đồ mối quan hệ giữa sales và newspaper (biểu đồ 3). /content/drive/MyDrive/Học máy cơ bản/Practice05/Advertising.csv
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# 1. Đọc dữ liệu từ file Advertising.csv
df = pd.read_csv('/content/drive/MyDrive/Học máy cơ bản/Practice05/Advertising.csv')

# Khởi tạo khung vẽ gồm 3 biểu đồ nằm ngang
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 5))

# Biểu đồ 1: Mối quan hệ giữa sales và TV
sns.scatterplot(data=df, x='TV', y='sales', ax=ax1, color='blue')
ax1.set_title('Mối quan hệ giữa Sales và TV')
ax1.grid(True)

# Biểu đồ 2: Mối quan hệ giữa sales và radio
sns.scatterplot(data=df, x='radio', y='sales', ax=ax2, color='green')
ax2.set_title('Mối quan hệ giữa Sales và Radio')
ax2.grid(True)

# Biểu đồ 3: Mối quan hệ giữa sales và newspaper
sns.scatterplot(data=df, x='newspaper', y='sales', ax=ax3, color='orange')
ax3.set_title('Mối quan hệ giữa Sales và Newspaper')
ax3.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Yêu cầu 2: Đọc dữ liệu từ file Advertising.csv. Mã hoá các cột Text (nếu có) và
# chia dữ liệu random thành 2 phần: 2/3 training và phần còn lại cho testing.
from sklearn.model_selection import train_test_split

# 1. Tách các cột đặc trưng (X) và nhãn mục tiêu (y)
# Loại bỏ cột đầu tiên (index) và cột mục tiêu 'sales'
X = df[['TV', 'radio', 'newspaper']].values
y = df['sales'].values

# 2. Chia dữ liệu theo tỷ lệ 2/3 (khoảng 66.67%) cho training và 1/3 (33.33%) cho testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=1/3, random_state=42)

print("Số lượng mẫu huấn luyện (Train):", len(X_train))
print("Số lượng mẫu kiểm thử (Test):", len(X_test))
```

```python
# Yêu cầu 3: Huấn luyện, đánh giá và hiển thị giá trị MAE, MSE, R2 với giải thuật Linear Regression.
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (mean_squared_error, mean_absolute_error, r2_score)
# 1. Xử lý dataset: ở trên
# 2. Khởi tạo và huấn luyện mô hình Linear Regression
regressor_1 = LinearRegression()
regressor_1.fit(X_train, y_train)
# 3. Dự đoán
y_pred = regressor_1.predict(X_test)
# 4. Đánh giá mô hình
mse_1 = mean_squared_error(y_test, y_pred)
mae_1 = mean_absolute_error(y_test,y_pred)
r2_1 = r2_score(y_test, y_pred)
print("Mean Square Error (MSE): ", mse_1)
print("Mean Absolute Error (MAE): ", mae_1)
print("R2 Score:", r2_1)
```

```python
# Yêu cầu 4: Huấn luyện, đánh giá và hiển thị giá trị MAE, MSE, R2 với giải thuật Decision Tree.
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import (mean_squared_error, mean_absolute_error, r2_score)

# 1. Khởi tạo và huấn luyện mô hình
regressor_2 = DecisionTreeRegressor(random_state=0)
regressor_2.fit(X_train, y_train)

# 2. Dự đoán
y_pred = regressor_2.predict(X_test)

# 3. Đánh giá mô hình
mse_2 = mean_squared_error(y_test, y_pred)
mae_2 = mean_absolute_error(y_test, y_pred)
r2_2 = r2_score(y_test, y_pred)

# 4. Hiển thị kết quả
print("Mean Squared Error (MSE):", mse_2)
print("Mean Absolute Error (MAE):", mae_2)
print("R2 Score:", r2_2)
```

```python
# Yêu cầu 5: Huấn luyện, đánh giá và hiển thị giá trị MAE, MSE, R2 với giải thuật Random Forest.
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (mean_squared_error, mean_absolute_error, r2_score)

# 1. Khởi tạo và huấn luyện mô hình
regressor_3 = RandomForestRegressor(random_state=0)
regressor_3.fit(X_train, y_train)

# 2. Dự đoán
y_pred = regressor_3.predict(X_test)

# 3. Đánh giá mô hình
mse_3 = mean_squared_error(y_test, y_pred)
mae_3 = mean_absolute_error(y_test, y_pred)
r2_3 = r2_score(y_test, y_pred)

# 4. Hiển thị kết quả
print("Mean Squared Error (MSE):", mse_3)
print("Mean Absolute Error (MAE):", mae_3)
print("R2 Score:", r2_3)
```

```python
# Yêu cầu 6: Huấn luyện, đánh giá và hiển thị giá trị MAE, MSE, R2 với giải thuật Gradient Boosting.
import matplotlib.pyplot as plt
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import (mean_squared_error, mean_absolute_error, r2_score)

# 1. Khởi tạo và huấn luyện mô hình
regressor_4 = GradientBoostingRegressor(random_state=0)
regressor_4.fit(X_train, y_train)

# 2. Dự đoán
y_pred = regressor_4.predict(X_test)

# 3. Đánh giá mô hình
mse_4 = mean_squared_error(y_test, y_pred)
mae_4 = mean_absolute_error(y_test, y_pred)
r2_4 = r2_score(y_test, y_pred)

# 4. Hiển thị kết quả
print("Mean Squared Error (MSE):", mse_4)
print("Mean Absolute Error (MAE):", mae_4)
print("R2 Score:", r2_4)
```

```python
# Yêu cầu 7: Huấn luyện, đánh giá và hiển thị giá trị MAE, MSE, R2 với giải thuật XGBoost.
import matplotlib.pyplot as plt
from xgboost import XGBRegressor
from sklearn.metrics import (mean_squared_error, mean_absolute_error, r2_score)

# 1. Khởi tạo và huấn luyện mô hình
regressor_5 = XGBRegressor(random_state=0)
regressor_5.fit(X_train, y_train)

# 2. Dự đoán
y_pred = regressor_5.predict(X_test)

# 3. Đánh giá mô hình
mse_5 = mean_squared_error(y_test, y_pred)
mae_5 = mean_absolute_error(y_test, y_pred)
r2_5 = r2_score(y_test, y_pred)

# 4. Hiển thị kết quả
print("Mean Squared Error (MSE):", mse_5)
print("Mean Absolute Error (MAE):", mae_5)
print("R2 Score:", r2_5)
```

```python
# Yêu cầu 8: Thực hiện dự đoán với từng giải thuật ở trên.
import numpy as np

# 1. Tạo một mẫu dữ liệu chi phí quảng cáo mới cụ thể
# Định dạng: [TV, radio, newspaper]
new_sample = np.array([[200.0, 100.0, 50.0]])

# 2. Tiến hành dự đoán với từng mô hình đã huấn luyện ở trên
# Lưu kết quả vào biến mới để tránh ghi đè làm mất đối tượng mô hình gốc
pred_1 = regressor_1.predict(new_sample)[0]
pred_2 = regressor_2.predict(new_sample)[0]
pred_3 = regressor_3.predict(new_sample)[0]
pred_4 = regressor_4.predict(new_sample)[0]
pred_5 = regressor_5.predict(new_sample)[0]

# 3. In kết quả dự đoán một cách trực quan
print(f"Thông số chi phí quảng cáo nhập vào: TV={new_sample[0][0]}, Radio={new_sample[0][1]}, Newspaper={new_sample[0][2]}\n")
print("--- KẾT QUẢ DỰ ĐOÁN DOANH SỐ (SALES) ---")
print(f"1. Linear Regression:  {pred_1:.4f}")
print(f"2. Decision Tree:     {pred_2:.4f}")
print(f"3. Random Forest:     {pred_3:.4f}")
print(f"4. Gradient Boosting: {pred_4:.4f}")
print(f"5. XGBoost:           {pred_5:.4f}")
```
