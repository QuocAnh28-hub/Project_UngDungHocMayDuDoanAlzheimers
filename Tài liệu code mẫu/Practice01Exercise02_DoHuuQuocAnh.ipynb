```python
import pandas as pd

path = '/content/drive/MyDrive/Học máy cơ bản/101Traffic_accident.csv'

df = pd.read_csv(path)


#1: Thống kê xem có bao nhiêu loại phương tiện bị tai nạn (Vehicle_type).
x = (df['Vehicle_type'].unique())
print("Số loại hương tiện bị tai nạn: ", len(x))
```

```python
#2: Hiển thị số vụ tai nạn theo từng loại phương tiện.
df['Vehicle_type'].value_counts()
```

```python
#3: Hiển thị số nạn nhân tử vong (death_toll) theo từng tháng.
df.groupby('Month')['death_toll'].sum()
```

```python
#4: Hiển thị số nạn nhân bị thương (Injuries_Person) theo tháng.
df.groupby('Month')['Injuries_Person'].sum()
```

```python
#5: Hiển thị số vụ tai nạn theo từng tháng (Month).
df['Month'].value_counts().sort_index()
```

```python
#6: Có bao nhiêu nạn nhân bị thương dưới 18 tuổi?
print('Số nạn nhân bị thương dưới 18 tuổi: ', df[df['Age'] < 18]['Injuries_Person'].sum())
```

```python
#7: Có bao nhiêu nạn nhân tử vong trên 18 tuổi.
print('Số nạn nhân tử vong trên 18 tuổi: ', df[df['Age'] > 18]['death_toll'].sum())
```

```python
#8: Thống kê số nạn nhân nữ và nam bị tai nạn năm đó.
df['Sex'].value_counts()
```

```python
#9: Thống kê và hiển thị tỉ lệ nạn nhân nữ và nam bị tai nạn.
x = (df['Sex'] == 1).sum()
y = (df['Sex'] == 2).sum()
z = (df['Sex'] == 3).sum()
tong = x + y + z
x = (x / tong) * 100
y = (y / tong) * 100
print("Tỷ lệ % nam bị tai nạn: ", x, "%")
print("Tỷ lệ % nữ bị tai nạn: ", y, "%")
```

```python
#10: Có bao nhiêu nạn nhân là nữ trên 60 tuổi.
print('Số nạn nhân là nữ trên 60 tuổi: ', len(df[(df['Age'] > 60) & (df['Sex']==2)]))
```

```python
#11: Hiển thị bar chart (biểu đồ cột) số vụ tai nạn theo tháng.
import matplotlib.pyplot as plt
vu_theo_thang = df['Month'].value_counts().sort_index()

# Vẽ biểu đồ cột
vu_theo_thang.plot(kind='bar')

plt.title('Số vụ tai nạn theo tháng')
plt.xlabel('Tháng')
plt.ylabel('Số vụ tai nạn')
plt.xticks(rotation=0)
plt.show()
```

```python
#12: Hiển thị line chart và bar chart số nạn nhân tử vong theo tháng.
import matplotlib.pyplot as plt
nan_nhan_tu_vong_theo_thang = df.groupby(df['Month'])['death_toll'].sum()

# Vẽ biểu đồ cột
nan_nhan_tu_vong_theo_thang.plot(kind='bar')

plt.title('Số nạn nhân tử vong theo tháng - Bar Chart')
plt.xlabel('Tháng')
plt.ylabel('Số nạn nhân tử vong')
plt.xticks(rotation=0)
plt.show()

# Vẽ biểu đồ đường
nan_nhan_tu_vong_theo_thang.plot(kind='line', marker='o')

plt.title('Số nạn nhân tử vong theo tháng - Line Chart')
plt.xlabel('Tháng')
plt.ylabel('Số nạn nhân tử vong')
plt.xticks(rotation=0)
plt.show()
```
