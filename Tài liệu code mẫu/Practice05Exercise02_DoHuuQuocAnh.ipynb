```python
# Yêu cầu 1: Đọc dữ liệu từ file insurance.csv. Mã hoá các cột Text và chia dữ liệu
# random thành 2 phần: 2/3 training và phần còn lại cho testing.

# 1. XỬ LÝ DỮ LIỆU
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
# 1.1. Đọc file CSV
dataset = pd.read_csv('/content/drive/MyDrive/Học máy cơ bản/Practice05/insurance.csv')

# 1.2. Mã hóa các cột Text
dataset = pd.get_dummies(dataset, columns=['sex', 'smoker', 'region'], dtype=int)

# 1.3. Tách dữ liệu đầu vào X và đầu ra y
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values

# 1.4. Chia dữ liệu thành training và testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=1/3, random_state=0)

# 1.5. Hiển thị số bản ghi
print("Number of Training samples:", len(X_train))
print("Number of Testing samples:", len(X_test))
```

```python
# Yêu cầu 2: Huấn luyện, đánh giá và hiển thị giá trị MAE, MSE, R2 với giải thuật Linear Regression.
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (mean_squared_error, mean_absolute_error, r2_score)
# 1. Xử lý dataset: ở trên
# 2. Khởi tạo và huấn luyện mô hình Linear Regression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
# 3. Dự đoán
y_pred = regressor.predict(X_test)
# 4. Đánh giá mô hình
mse_1 = mean_squared_error(y_test, y_pred)
mae_1 = mean_absolute_error(y_test,y_pred)
r2_1 = r2_score(y_test, y_pred)
print("Mean Square Error (MSE): ", mse_1)
print("Mean Absolute Error (MAE): ", mae_1)
print("R2 Score:", r2_1)
```

```python
# Yêu cầu 3: Huấn luyện, đánh giá và hiển thị giá trị MAE, MSE, R2 với giải thuật Decision Tree.
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import (mean_squared_error, mean_absolute_error, r2_score)

# 1. Khởi tạo và huấn luyện mô hình
regressor = DecisionTreeRegressor(random_state=0)
regressor.fit(X_train, y_train)

# 2. Dự đoán
y_pred = regressor.predict(X_test)

# 3. Đánh giá mô hình
mse_2 = mean_squared_error(y_test, y_pred)
mae_2 = mean_absolute_error(y_test, y_pred)
r2_2 = r2_score(y_test, y_pred)

# 4. Hiển thị kết quả
print("Mean Squared Error (MSE):", mse_2)
print("Mean Absolute Error (MAE):", mae_2)
print("R2 Score:", r2_2)
```

```python
# Yêu cầu 4: Huấn luyện, đánh giá và hiển thị giá trị MAE, MSE, R2 với giải thuật Random Forest.
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (mean_squared_error, mean_absolute_error, r2_score)

# 1. Khởi tạo và huấn luyện mô hình
regressor = RandomForestRegressor(random_state=0)
regressor.fit(X_train, y_train)

# 2. Dự đoán
y_pred = regressor.predict(X_test)

# 3. Đánh giá mô hình
mse_3 = mean_squared_error(y_test, y_pred)
mae_3 = mean_absolute_error(y_test, y_pred)
r2_3 = r2_score(y_test, y_pred)

# 4. Hiển thị kết quả
print("Mean Squared Error (MSE):", mse_3)
print("Mean Absolute Error (MAE):", mae_3)
print("R2 Score:", r2_3)
```

```python
# Yêu cầu 5: Huấn luyện, đánh giá và hiển thị giá trị MAE, MSE, R2 với giải thuật Gradient Boosting.
import matplotlib.pyplot as plt
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import (mean_squared_error, mean_absolute_error, r2_score)

# 1. Khởi tạo và huấn luyện mô hình
regressor = GradientBoostingRegressor(random_state=0)
regressor.fit(X_train, y_train)

# 2. Dự đoán
y_pred = regressor.predict(X_test)

# 3. Đánh giá mô hình
mse_4 = mean_squared_error(y_test, y_pred)
mae_4 = mean_absolute_error(y_test, y_pred)
r2_4 = r2_score(y_test, y_pred)

# 4. Hiển thị kết quả
print("Mean Squared Error (MSE):", mse_4)
print("Mean Absolute Error (MAE):", mae_4)
print("R2 Score:", r2_4)
```

```python
# Yêu cầu 6: Huấn luyện, đánh giá và hiển thị giá trị MAE, MSE, R2 với giải thuật XGBoost.
import matplotlib.pyplot as plt
from xgboost import XGBRegressor
from sklearn.metrics import (mean_squared_error, mean_absolute_error, r2_score)

# 1. Khởi tạo và huấn luyện mô hình
regressor = XGBRegressor(random_state=0)
regressor.fit(X_train, y_train)

# 2. Dự đoán
y_pred = regressor.predict(X_test)

# 3. Đánh giá mô hình
mse_5 = mean_squared_error(y_test, y_pred)
mae_5 = mean_absolute_error(y_test, y_pred)
r2_5 = r2_score(y_test, y_pred)

# 4. Hiển thị kết quả
print("Mean Squared Error (MSE):", mse_5)
print("Mean Absolute Error (MAE):", mae_5)
print("R2 Score:", r2_5)
```

```python
# Yêu cầu 7: So sánh với kết quả các giải thuật, xem giải thuật nào cho kết quả tốt nhất.
import pandas as pd

# Tổng hợp kết quả đánh giá của 3 mô hình
results = pd.DataFrame({
    'Algorithm': ["Linear Regression", "Decision Tree", "Random Forest", "Gradient Boosting", "XGBoost"],
    'MAE': [mae_1, mae_2, mae_3, mae_4, mae_5],
    'MSE': [mse_1, mse_2, mse_3, mse_4, mse_5],
    'R2': [r2_1, r2_2, r2_3, r2_4, r2_5]
})

# Hiển thị bảng so sánh
print(results)

# Xác định mô hình tốt nhất theo từng tiêu chí
print("\nBest model according to MAE:",
      results.loc[results['MAE'].idxmin(), 'Algorithm'])

print("Best model according to MSE:",
      results.loc[results['MSE'].idxmin(), 'Algorithm'])

print("Best model according to R2:",
      results.loc[results['R2'].idxmax(), 'Algorithm'])
```
